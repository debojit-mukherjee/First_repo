```python
import requests # Import requests library for API calls

# Stage 3: Implementation with Exercises + Feedback

def greet_user(name):
    return f"Hello {name}, I am your AI Chatbot! I can teach SQL, Python, Java, C++, and HTML."

def choose_topic():
    topic = input("Which topic would you like to learn? (SQL/Python/Java/C++/HTML/API): ")
    return topic

def explain_topic(topic):
    if topic.lower() == "python":
        return "Python is beginner-friendly. Let's start with variables!"
    elif topic.lower() == "sql":
        return "SQL is used to manage databases. Let's start with SELECT queries!"
    elif topic.lower() == "java":
        return "Java is powerful for applications. Let's start with classes!"
    elif topic.lower() == "c++":
        return "C++ is great for performance-heavy apps. Let's start with functions!"
    elif topic.lower() == "html":
        return "HTML is the backbone of web pages. Let's start with headings!"
    elif topic.lower() == "api":
        print("APIs let applications talk to each other. Let's start with GET requests!")
        print("Let's try a real API call!")
        response = requests.get("https://jsonplaceholder.typicode.com/posts/1")
        if response.status_code == 200:
            data = response.json()
            print("API Response Example:")
            print(f"Title: {data['title']}")
            print(f"Body: {data['body']}")
        else:
            print("API call failed.")
        return ""
    else:
        return "Sorry, I don’t know that topic yet."

def practice_exercise(topic):
    if topic.lower() == "python":
        import random
        python_exercises = [
            ("What will this print?\nprint(2*3)", "6"),
            ("What keyword defines a function?", "def"),
            ("What symbol is used for comments?", "#")
        ]

        question, correct_answer = random.choice(python_exercises)
        answer = input(f"Exercise: {question}\nYour answer: ")
        if answer.strip().lower() == correct_answer.lower():
          print("Correct! 🎉")
        else:
          print(f"Try again! The answer is {correct_answer}.")


    elif topic.lower() == "sql":
        answer = input("Exercise: Which SQL command retrieves data?\nYour answer: ")
        if answer.lower().strip() == "select":
            print("Correct! 🎉")
        else:
            print("Try again! The answer is SELECT.")

    elif topic.lower() == "java":
        answer = input("Exercise: What keyword is used to define a class in Java?\nYour answer: ")
        if answer.lower().strip() == "class":
            print("Correct! 🎉")
        else:
            print("Try again! The answer is 'class'.")

    elif topic.lower() == "c++":
        answer = input("Exercise: Which symbol is used for single-line comments in C++?\nYour answer: ")
        if answer.strip() == "//":
            print("Correct! 🎉")
        else:
            print("Try again! The answer is //.")

    elif topic.lower() == "html":
        answer = input("Exercise: Which tag is used for the largest heading in HTML?\nYour answer: ")
        if answer.lower().strip() == "<h1>":
            print("Correct! 🎉")
        else:
            print("Try again! The answer is <h1>.")

    elif topic.lower() == "api":
        answer = input("Which HTTP method did we just use? (GET/POST/PUT/DELETE): ")
        if answer.lower().strip() == "get":
            print("Correct! 🎉 We used GET to retrieve data.")
        else:
            print("Not quite — the correct answer is GET.")

        # --- POST Request Demo ---
        print("\nNow let's try sending data with POST...")
        new_post = {
            "title": "Debojit's Chatbot Lesson",
            "body": "Learning how POST works!",
            "userId": 1
        }
        response = requests.post("https://jsonplaceholder.typicode.com/posts", json=new_post)

        if response.status_code == 201:  # 201 means 'Created'
            data = response.json()
            print("POST Response Example:")
            print(f"ID: {data['id']}")
            print(f"Title: {data['title']}")
            print(f"Body: {data['body']}")
        else:
            print("POST request failed.")

        answer = input("Which HTTP method did we just use to send new data? (GET/POST/PUT/DELETE): ")
        if answer.lower().strip() == "post":
            print("Correct! 🎉 We used POST to send new data.")
        else:
            print("Not quite — the correct answer is POST.")


        # --- PUT Request Demo ---
        print("\nNow let's try updating data with PUT...")
        updated_post = {
            "id": 1,
            "title": "Updated Chatbot Lesson",
            "body": "This shows how PUT updates data!",
            "userId": 1
        }
        response = requests.put("https://jsonplaceholder.typicode.com/posts/1", json=updated_post)

        if response.status_code == 200:  # 200 means 'OK'
            data = response.json()
            print("PUT Response Example:")
            print(f"ID: {data['id']}")
            print(f"Title: {data['title']}")
            print(f"Body: {data['body']}")
        else:
            print("PUT request failed.")

        answer = input("Which HTTP method did we just use to update data? (GET/POST/PUT/DELETE): ")
        if answer.lower().strip() == "put":
            print("Correct! 🎉 We used PUT to update data.")
        else:
            print("Not quite — the correct answer is PUT.")

        # --- DELETE Request Demo ---
        print("\nFinally, let's try removing data with DELETE...")
        response = requests.delete("https://jsonplaceholder.typicode.com/posts/1")

        if response.status_code == 200:  # 200 means 'OK'
            print("DELETE Response Example:")
            print("Post deleted successfully (fake API, so nothing really removed).")
        else:
            print("DELETE request failed.")

        # --- Interactive Question ---
        answer = input("Which HTTP method did we just use to remove data? (GET/POST/PUT/DELETE): ")
        if answer.lower().strip() == "delete":
            print("Correct! 🎉 We used DELETE to remove data.")
        else:
            print("Not quite — the correct answer is DELETE.")

    else:
        print("Exercises for this topic are coming soon!")

# Stage 4 + 5: Run interactively in Colab
def chatbot():
    name = input("Enter your name: ")
    print(greet_user(name))

    topic = choose_topic()
    explain_topic(topic)

    practice_exercise(topic)

# Start the chatbot
chatbot()
```
